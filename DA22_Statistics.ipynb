# Operations on matrix

### 1. Matrix Addition and Subtraction

```python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A + B)
print(A - B)
```

```python
## 2. Scaler multiplication
scaler  = 10
A = np.array([[1,2],
              [3,4]])

print(scaler * A)
```

```python
## 3 Dot product on vectors
a = np.array([1,2])
b = np.array([3,4])

print(a @ b)
print(np.dot(a,b))
```

```python
# 4. Dot product on matrics
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A @ B)
```

```python
# transpose of matrix
# row <-> column

A = np.array([[10,20,30,40],
              [50,60,70,80]])

print(A.T)
```

```python
# Determinant of a matrix
A = np.array([[1,2],
              [3,4]])

print(int(np.linalg.det(A)))
```

```python
# Inverse of matrix
A = np.array([[1,2],[3,4]])
print(np.linalg.inv(A))
```

# Measure of Dispersion

```python
# Spreadness of data : how much the data is far away from mean/median
# MAD, Variance, Standard deviation, Percentile, Quantile, IQR, five number summary
```

```python
import numpy as np

# Mean Absolute deviation (MAD)

Reelcount = [50, 10, 15, 20, 50, 10, 10, 11]

mean = np.mean(Reelcount)

int_result = []

for i in Reelcount:
    int_result.append((np.abs(i - mean)))

MAD = (1/len(Reelcount)) * np.sum(int_result)
print(MAD)
```

```python
# variance
var = np.var(Reelcount)
print(var)
```

```python
# Standard deviation
print(np.std(Reelcount))
```

```python
SD = np.sqrt(var)
print(SD)
```

```python
# Percentile
Reelcount = [50, 10, 15, 20, 50, 70, 80, 11, 5, 2]
Q1 = np.percentile(Reelcount, 25)
Q2 = np.percentile(Reelcount, 50)
Q3 = np.percentile(Reelcount, 75)
IQR = Q3 - Q1
LB = Q1 - 1.5 * IQR
UB = Q3 + 1.5 * IQR
print(Q1, Q3, IQR, UB, LB)
```

# 2.1.1 Simple sampling technique

# Confidance  interval / Interval Estimation

```python
import numpy as np
from scipy.stats import norm
# CI when SD is Known (using Z Table)
# Question: find a interval range with 95% confidance
Paneer_pieces = [ 10, 5, 4, 10, 8, 0, 2]
# CL = 95%, alpha = 0.05, alpha/2 = 0.025, 1-0.025 = 0.975

n = len(Paneer_pieces)

x_bar = np.mean(Paneer_pieces)
pop_sd = np.std(Paneer_pieces, ddof = 0)
sqrt_n = np.sqrt(n)

z_value = norm.ppf(0.975)

inter_value = z_value * (pop_sd / sqrt_n)

CI_Z = (x_bar  - inter_value, x_bar + inter_value)
CI_Z
```

```python
# Confidance interval when SD is Unknown using t table
import numpy as np
from scipy.stats import t

# Question: find a interval range with 95% confidance
Paneer_pieces = [ 10, 5, 4, 10, 8, 0, 2]
# CL = 95%, alpha = 0.05, alpha/2 = 0.025, 1-0.025 = 0.975

n = len(Paneer_pieces)

x_bar = np.mean(Paneer_pieces)
sample_sd = np.std(Paneer_pieces, ddof = 1)
sqrt_n = np.sqrt(n)

t_value = t.ppf(0.975, n-1)

inter_value = t_value * (sample_sd / sqrt_n)

CI_t = (x_bar - inter_value, x_bar + inter_value)
CI_t
```

# Hypothesis testing

# t-test

# One Sample t-test

```python
# Null Hypothesis (Claim) : Sample mean is approximatly same as population mean
# Alternate Hypothesis : Sample mean is significanlty different from population mean

import numpy as np
from scipy.stats import ttest_1samp


pop_mean = 50
samples = np.array([87, 88, 86, 89, 90, 84, 84, 88, 89, 90])
print(np.mean(samples))

t_statistics, p_value = ttest_1samp(samples, popmean = pop_mean)

alpha = 0.05

if p_value < alpha:
    print("Reject Null hypothesis")
else:
    print("Fail to Reject Null hypothesis")

# p_value is helping to find if reject or fail to reject null hypothesis based upon the alpha value 
```

# two sample t-test

```python
# Null hypothesis : Male salary is approximatly same as Female salary
# Alternate hypothesis: Male salary is significalty different from the Female salary

from scipy.stats import ttest_ind

np.random.seed(42)
male_salary = np.random.randint(200, 400, size = 29)
print(np.mean(male_salary))

female_salary = np.random.randint(200, 400, size = 29)
print(np.mean(female_salary))

t_statistics, p_value = ttest_ind(male_salary, female_salary)

alpha = 0.05
if p_value < alpha:
    print("Reject Null hypothesis")
else:
    print("Fail to Reject Nul hypothesis")
```

# paired t test

```python
# Null Hypothesis: No significant difference between Before and After
# Alterante : Their is significant difference between Before and After

# Null hypothesis: No significant difference in your weight between before and after joining the PG
# Alernate : Their is significant difference in weight

from scipy.stats import ttest_rel
import numpy as np

np.random.seed(42)
before = np.random.randint(50, 70, size = 29)
after = np.random.randint(150, 170, size = 29)

t_statistics, p_value = ttest_rel(before, after)  # two tailed test

alpha = 0.05
if p_value < alpha:
    print("Reject Null hypothesis")
else:
    print("Fail to reject null hypothesis")
```
