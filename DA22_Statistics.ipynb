# Linear Algebra

### Scaler, Vector, Matrix and Tensor

```python
import numpy as np
```

```python
# scaler : 0D array

salary = np.array(500000)
print(salary)
print(salary.ndim)
print(type(salary))
```

```python
# vector : 1D array
expense = np.array([6000, 4000, 10000000000, 2000])
print(expense)
print(expense.ndim)
print(type(expense))
print(expense.shape)
```

```python
# matrix : 2D array
expense = np.array([[6000, 4000, 10000000000, 2000],
                    [20000, 30000, 500000, 100000]])
print(expense)
print(expense.ndim)
print(type(expense))
print(expense.shape)
```

```python
# tensor : 3D array
expense = np.array([[[6000, 4000, 10000000000, 2000],
                    [20000, 30000, 500000, 100000]],
                  
                  [[6000, 4000, 10000000000, 2000],
                    [20000, 30000, 500000, 100000]]])
print(expense)
print(expense.ndim)
print(type(expense))
print(expense.shape)
```

```python
# tensor : 4D array
expense = np.array([[[[6000, 4000, 10000000000, 2000],
                    [20000, 30000, 500000, 100000]],
                  
                  [[6000, 4000, 10000000000, 2000],
                    [20000, 30000, 500000, 100000]]],
                  
                  [[[6000, 4000, 10000000000, 2000],
                    [20000, 30000, 500000, 100000]],
                  
                  [[6000, 4000, 10000000000, 2000],
                    [20000, 30000, 500000, 100000]]]])
print(expense)
print(expense.ndim)
print(type(expense))
print(expense.shape)
```

# Operations on matrix

### 1. Matrix Addition and Subtraction

```python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A + B)
print(A - B)
```

```python
## 2. Scaler multiplication
scaler  = 10
A = np.array([[1,2],
              [3,4]])

print(scaler * A)
```

```python
## 3 Dot product on vectors
a = np.array([1,2])
b = np.array([3,4])

print(a @ b)
print(np.dot(a,b))
```

```python
# 4. Dot product on matrics
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A @ B)
```

```python
# transpose of matrix
# row <-> column

A = np.array([[10,20,30,40],
              [50,60,70,80]])

print(A.T)
```

```python
# Determinant of a matrix
A = np.array([[1,2],
              [3,4]])

print(int(np.linalg.det(A)))
```

```python
# Inverse of matrix
A = np.array([[1,2],[3,4]])
print(np.linalg.inv(A))
```
