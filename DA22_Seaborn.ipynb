# line chart

# All the charts of matplotlib can also be plotted by seaborn

```python
# trends over the time

# on the same chart features of seaborn and matplolib can be apply

import matplotlib.pyplot as plt

timeline = df.groupby("year")["passengers"].sum().reset_index()

sns.lineplot(x = "year", y = "passengers", data = timeline, marker = "o" )
plt.title("Year wise passengers")
plt.grid()
```

```python
# Scatter plot

df = sns.load_dataset("tips")
df
```

```python
# to find the relation between two varaibles

plt.figure(figsize = (3,2))

sns.scatterplot(x = "total_bill", y = "tip", data = df, hue = "smoker") 
```

```python
# pair plot : compare all possible numerical pair of variables at a time

plt.figure(figsize = (4,3))
# compare multiple columns with each other together
sns.pairplot(df, hue = "smoker", kind='scatter')
```

# Heatmap

```python
# USE: heatmap visualize the EXACT correlation between numerical/categorical variables

# Heatmap is used for ploting correlation matrix in colorful chart

import pandas as pd

# df.corr() function is used to generate correlation table
corr = df.corr(numeric_only = True)

print(corr)
```

```python
sns.heatmap(corr, annot = True, cmap = "YlGnBu", fmt = ".3f")
```

# Heatmap for categorical variable 

```python
# Finding relation between categorical variable

# numerical relation : heatmap + df.corr()

# categorical relation : heatmap + Pivot table

# df.corr() function is used for finding relation between numerical variables
# Pivot table is used for finding relation between categorical variables.

# Heatmap is used just to show the relation tables in colorful chart
```

```python
pivot = df.pivot_table(index = "day", columns = "sex", aggfunc = "size", observed=False)

sns.heatmap(pivot, annot = True, cmap = "coolwarm")
```

```python
# Example-2

# size here is not the column name
# size : calculate total elements category wise
pivot = df.pivot_table(index = "day", columns = "smoker", values = "total_bill", aggfunc = "mean", observed=False)

plt.figure(figsize = (3,2))
sns.heatmap(pivot, annot = True)
```

# countplot

```python
# USE: shows the frequency of different categories

sns.countplot(data = df, x = 'time', hue = "sex")
```

# Box plot

```python
# identify the outliers

# outlier : the data which is not matching with most of the other data
```

```python
plt.figure(figsize = (3,2))
sns.boxplot(x = "tip", data = df)
```

```python
plt.figure(figsize = (5,4))
sns.boxplot(x = "day", y = "tip", data = df, hue = "smoker")
```

```python
sns.boxplot(data = df)
```

# KDE (Kernel Density Estimate plot) / Density plot

```python
# USE: KDE plot shows the density of data range wise
```

```python
plt.figure(figsize = (3,2))
sns.kdeplot(data = df, x = "total_bill", fill = True)
```

# violin plot

```python
# violin plot = Box plot + Density plot

sns.violinplot(x = "smoker", y = "total_bill", data = df)
```

# Regression plot

```python
# USE: used to plot the linear regression line
plt.figure(figsize = (3,2))
sns.regplot(x = "total_bill", y = "tip", data = df)
```

# facetgrid

```python
# USE: to plot multiple charts based upon classes in categorical variable
```

```python
plt.figure(figsize = (2,1))
obj = sns.FacetGrid(df, col = "day")
obj.map_dataframe(sns.scatterplot, x = "total_bill", y = "tip")

plt.show()
```
