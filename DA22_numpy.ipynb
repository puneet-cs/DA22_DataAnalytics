# Modifications, insertion and deletion in numpy

```python
import numpy as np
south_sweet = np.array([["Motichoor", "Kalakand", "Jalebi"],
                        ["Bobbatlu", "Kaji Kaya", "Khaja"]])

# Modification of single element
south_sweet[0,0] = 'Payasum'
print(south_sweet)
```

```python
# updated complete row
south_sweet[1] = ['Kesari', 'Pulareddy', 'arasalu']
print(south_sweet)
```

```python
# update the complete column
# numpy array take fix space to store data
south_sweet[:, 2] = ['Mysurepak', 'Doublekamitha']
print(south_sweet)
```

```python
# Insert operation : New row

south_sweet = np.insert(south_sweet, 1, ['Gajarpak', 'Rusmali', 'Firni'], axis = 0)
print(south_sweet)
```

```python
# Insert operation : New column

south_sweet = np.insert(south_sweet, 2, ['rasbhari', 'Gulabjamun', 'Kheer'], axis = 1)
print(south_sweet)
```

```python
# Insert operation : multiple columns

south_sweet = np.insert(south_sweet, 2, 
                        [['Kajukatli', 'Ghevar', 'Peda'],['Kajukatli', 'Ghevar', 'Peda'] ], 
                        axis = 1)
print(south_sweet)
```

```python
# Delete operation 
south_sweet = np.delete(south_sweet, 0, axis = 0)
print(south_sweet)
```

```python
# Delete operation : column
south_sweet = np.delete(south_sweet, 2, axis = 1)
print(south_sweet)
```

```python
# Delete operation : column
south_sweet = np.delete(south_sweet, [1,3], axis = 1)
print(south_sweet)
```

# 8. Concatenation, Stacking, Splitting numpy arrays

```python
# concatenate the 1d's array's

arr1 = np.array([100, 200, 300])
arr2 = np.array([90, 150, 5])
arr3 = np.array([10, 20, 30])

new_arr = np.concatenate((arr1, arr2, arr3))
print(new_arr)
```

```python
# concatenate the 2D arrays

arr1 = np.array([[10, 20, 30],
                 [40, 50, 60]])

arr2 = np.array([[100, 200, 300],
                 [400, 500, 600]])

# Vertical concatenation : Row wise
new_2d_arr = np.concatenate((arr1, arr2), axis = 0)
print(new_2d_arr)
```

```python
# Horizontal concatenation : Column wise
new_2d_arr = np.concatenate((arr1, arr2), axis = 1)
print(new_2d_arr)
```

```python
# We cannot concatenate 1D & 2D arrays
# We can vertically concatenate the 2D's array of different shape

arr1 = np.array([[100, 200, 300]])
arr2 = np.array([[100, 200, 300],
                 [400, 500, 600]])

# Vertical concatenation : Row wise
new_2d_arr = np.concatenate((arr1, arr2), axis = 0)
print(new_2d_arr)
```

```python
# Horizontally concatenation
new_2d_arr = np.concatenate((arr1, arr2), axis = 1)
print(new_2d_arr)
```

```python
arr1 = np.array([[100, 200],
                [300, 400]])
arr2 = np.array([[100, 200, 300],
                 [400, 500, 600]])

# Horizontally concatenation
new_2d_arr = np.concatenate((arr1, arr2), axis = 1)
print(new_2d_arr)
```

# np.stack() function

```python
# stack() function will add a "NEW DIMENSION" 
# Stack Vs Concatenate
# Concatenate don't add the new dimension while Stack add the new dimension
```

```python
arr1 = np.array([1000, 2000, 3000])
arr2 = np.array([300, 200, 100])

stacked_arr = np.stack((arr1, arr2))
print(stacked_arr)
```

```python
stacked_arr = np.stack((arr1, arr2), axis = 1)
print(stacked_arr)
print(stacked_arr.shape)
```

# vstack() : vertical stacking
# same as vertical concatenation

```python
arr1 = np.array([[10, 20, 30],
                 [40, 50, 60]])

arr2 = np.array([[100, 200, 300],
                 [400, 500, 600]])

vstack_arr = np.vstack((arr1, arr2))
print(vstack_arr)
```

# hstack() : horizontal stacking

#same as horizontal concatenation

```python
arr1 = np.array([[10, 20, 30],
                 [40, 50, 60]])

arr2 = np.array([[100, 200, 300],
                 [400, 500, 600]])

hstack_arr = np.hstack((arr1, arr2))
print(hstack_arr)
```
