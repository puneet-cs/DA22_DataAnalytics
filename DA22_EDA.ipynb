```python
df['education'].unique()
```

```python
df['occupation'].unique()
```

```python
df['occupation'] = df['occupation'].str.strip()
```

```python
df['occupation'].unique()
```

```python
df['occupation'] = df['occupation'].replace('?', 'other')
```

```python
df['occupation'].unique()
```

```python
df['relationship'].unique()
```

```python
df['relationship'] = df['relationship'].str.strip()
```

```python
df['relationship'].unique()
```

```python
df['race'].unique()
```

```python
df['race'] = df['race'].str.strip()
```

```python
df['native-country'].unique()
```

```python
df['native-country'] = df['native-country'].str.strip()
```

```python
df['native-country'] = df['native-country'].replace('?', 'other')
```

```python
df['sex'].unique()
```

```python
df.sex = df.sex.str.strip()
```

```python
df['income'].unique()
```

```python
df.income = df.income.str.strip()
```

```python
# interview Question: what is the difference between cov() & corr()?

# both shows the relation between numerical variables
# both show the direction of relation : positive, negative, no relation
# cov() will return -inf to +inf, corr() will return -1 to +1
# in corr() we can tell how much percent two variables are related to each other.
```
