### Merge two dataframes

```python
import pandas as pd
```

```python
# Def: Merge two dataframes based upon common column.
# functions used for merging dataframes : merge(), join(), concat()
```

```python
# merge() function
df1 = pd.DataFrame({'ID' : [1,2,3], 'DreamCompany' : ["Google", "Nvidia", "Accenture"], "Role" : ["DA", "DS", "TL"]})
df2 = pd.DataFrame({"ID" : [2,3,4], 'DreamPackage' : [20, 24, 50]})
print(df1)
print(df2)
```

```python
# Merge() function behaves like joins in SQL
# by deafult, merge() function perform inner join like operation
# inner merge : common records from both the data frames
merge_df = pd.merge(df1, df2, on = "ID")
merge_df
```

```python
merge_df = pd.merge(df1, df2, on = "ID", how = "inner")
merge_df
```

```python
# left merge : Pick all records from Left table and common records from the Right table
merge_df = pd.merge(df1, df2, on = "ID", how = "left")
merge_df
```

```python
# right merge : Pick all records from right table and common records from the left table
merge_df = pd.merge(df1, df2, on = "ID", how = "right")
merge_df
```

```python
# outer merge : Pick all records from right table and left table
# common records will be consider one time only
merge_df = pd.merge(df1, df2, on = "ID", how = "outer")
merge_df
```
