# 1. Data Structure creation (Series (1D array), DataFrame (2d array))

#### Series (1D array)

```python
# create series
ser1 = pd.Series()
print(ser1)
```

```python
# convert the list to Series
list1 = [10, 20, 30, 40, 50]
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))
```

```python
# convert the list to Series
list1 = [10, 20, 30, 40, 50,"Hello"]
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))

# In pandas strings are represented as object
# IMP: In Series, all elements have the same datatype
```

```python
# Example-2 : convert tuple to Series
tuple1 = (10, 20, 30, 40, 50)
ser2 = pd.Series(tuple1)
print(ser2)
print(type(ser2))
print(ser2[-2])

# In Pandas Series, direct negative indexing not allowed.
```

```python
# set cannot be converted to pandas series
# set is unordered and Series is ordered, so no conversion
set1 = {10, 20, 30, 40, 50}
ser2 = pd.Series(set1)
print(ser2)
```

```python
# Convert the numpy array into Series
import numpy as np

arr1 = np.array([10,20,30,40,50])
print(arr1, type(arr1))

ser3 = pd.Series(arr1, index = ['a', 'b', 'c', 'd', 'e'])
print(ser3, type(ser3))
print(ser3['a'])
```

```python
# update operation on series
ser2[3] = 100000

# Insert new value
ser2[5] = 2000
ser2['a'] = 5000
ser2[60] = 7000
ser2[3.4] = 9000
```

```python
print(ser2)
```

```python
# delete element from series
ser2 = ser2.drop(1)
ser2
```

# DataFrame (2d array)

```python
# Data frame is a 2D array, store data in form of matrix.
```

#### creating dataframe

```python
import pandas as pd
```

```python
df = pd.DataFrame()

print(df)
```

```python
# converting list to data frame
fruits_list = ["AratiPandu", "MamadiPandu", "KamalaPandu", "DrakshaPandu"]

df = pd.DataFrame(fruits_list, columns = ["Fruit Name"])
df
```

```python
# converting 2D list to DataFrame

fruits_list = [["AratiPandu",40], ["MamadiPandu", 1000], ["KamalaPandu", 10], ["DrakshaPandu",120]]

df = pd.DataFrame(fruits_list, columns = ["Fruit Name", "Price"])
df
```

```python
# created 3 columns dataframe (List of List)
fruits_list = [["AratiPandu",40,100], ["MamadiPandu", 1000, 200], ["KamalaPandu", 10, 1000], ["DrakshaPandu",120, -100]]

df = pd.DataFrame(fruits_list, columns = ["Fruit Name", "Price", "Discount"])
df
```

```python
# Create DataFrame from List of Tuple

fruits_list = [("AratiPandu",40,100), ("MamadiPandu", 1000, 200), ("KamalaPandu", 10, 1000), ("DrakshaPandu",120, -100)]

df = pd.DataFrame(fruits_list, columns = ["Fruit Name", "Price", "Discount"])
df
```

```python
# Convert the set of tuples into DataFrame

fruits_list = {("AratiPandu",40,100), ("MamadiPandu", 1000, 200), ("KamalaPandu", 10, 1000), ("DrakshaPandu",120, -100)}

df = pd.DataFrame(fruits_list, columns = ["Fruit Name", "Price", "Discount"])
df
```

```python
# Creating Dataframe from Set of Set

# In set, we cannot store the Mutable items
# In set, Only immutable items can be store

fruits_list = {["AratiPandu",40,100], ["MamadiPandu", 1000, 200], ["KamalaPandu", 10, 1000], ["DrakshaPandu",120, -100]}

df = pd.DataFrame(fruits_list, columns = ["Fruit Name", "Price", "Discount"])
df
```

```python
# Creating DataFrame from Dictionary

student_data = {
    "Name" : ["Shiva", "Niranjan", "Suresh", "Akash", "Lahari", "Rakesh"],
    "Salary" : [40, 50, 60, 15, 25, 100],
    "Study_time" : [12, 14, 16, 1, 5, 24]
}

df1 = pd.DataFrame(student_data)
df1
```
