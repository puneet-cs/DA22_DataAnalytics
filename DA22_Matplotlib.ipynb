# Basic plots : Bar chart, Histogram, Line chart, Scatter plot, Pie chart

# Bar chart

```python
# USE: to compare categorical and Discrete data
```

```python
food_category = ['Biryani', 'Idli', 'Dosa', 'Shawarma']
like_count = [50, 10, 20, 30]
plt.figure(figsize=(5,4))
plt.bar(food_category, like_count, color = ['gold', 'blue', 'green', 'orange'])

# to mention the value over the bar
for i, count in enumerate(like_count):
    plt.text(i, count+1, str(count), ha = 'center')

# i : x coordinate
# count+1 : y-coordinate
# str(count) : value to print on the chart
# ha = 'center' : horizontally placing value at the center of the bar

plt.xlabel("food categories")
plt.ylabel("Like count")
plt.title("food category like count")
```

# Horizontal bar chart

```python
# USE: when categories names are big then use the Horizontal bar chart

food_category = ['Biryani', 'Idli', 'Dosa', 'Shawarma', "Paneer chilly very tasty dum biryani"]
like_count = [50, 10, 20, 30, 40]

plt.barh(food_category, like_count)
```

# Histogram

```python
# USE: Histogram works on continuous data
# Used to show the frequency of data in a particular range

# Histogram is popularly used for checking the shape of data
```

```python
import numpy as np

data = np.random.randn(1000)   # random normal distributed data
plt.figure(figsize=(5,4))
plt.hist(data, bins = 10, edgecolor = 'black')
plt.ylabel("Frequency of data")
plt.xlabel("Continuous data")
plt.show()
```

```python
# Interview Question:
# What is the difference between Bar and Histogram charts?
# Ans:
# 1. Bar chart works on categorical/discrete data while histogram works on continus data
# 2. Bar chart is used for comparing categories while histogram is used for checking shape of data
```

# Line Chart

```python
# trends over the time
```

```python
import pandas as pd
import matplotlib.pyplot as plt

dates = pd.date_range(start = "2026-01-01", periods = 12, freq = 'ME')
icecream_sales = [91, 80, 100, 120, 130, 10, 20, 30, 40, 70, 10, 10]

plt.figure(figsize=(4,3))
plt.plot(dates, icecream_sales, marker = 'o')

for x,y in zip(dates, icecream_sales):
    plt.text(x,y, str(y))

plt.xticks(dates, dates.strftime("%b"), rotation = 45)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Icecream sales over the months")
plt.grid()
plt.show()
```

# Multiple lines chart

```python
# USE: compare two or more trends side by side
```

```python
dates = pd.date_range(start = "2015-01-01", periods = 12, freq = 'YE')

flipcart_sales = [90, 100, 10, 5, 20, 40, 45, 70, 90, 100, 80, 40]
amazon_sales = [100, 120, 130, 40, 30, 20, 20, 10, 5, 80, 100, 150]
Meesho_sales = [10, 12, 10, 40, 30, 20, 120, 10, 15, 80, 10, 100]

plt.plot(dates, flipcart_sales, marker = 'o', label = "FLIPCART")
plt.plot(dates, amazon_sales, marker = '*', label = "AMAZON")
plt.plot(dates, Meesho_sales, marker = '+', label = "MEESHO")

plt.legend()
```

# Pie chart

```python
# Pie chart is used on categorical data
# pie chart is used to show proportion of  the whole

# use pie chart when number of categories are less
```

```python
# mobile market share with pie chart

brands = ["Vivo", "Oppo", "Apple", "Sansung", "OnePlus", "Motorola", "Other"]
market_share = [20, 5, 30, 30, 5, 2, 8]
plt.figure(figsize = (5,4))
plt.pie(market_share, labels = brands, autopct = '%1.1f%%')

plt.savefig("pie1_da22.jpg")

plt.show()
```

# Donut chart

```python
# USE: Donut is like Pie chart with space in the middle
# This middle space can be further used to define more information
```

```python
brands = ["Vivo", "Oppo", "Apple", "Sansung", "OnePlus", "Motorola", "Other"]
market_share = [20, 5, 30, 30, 5, 2, 8]
plt.figure(figsize = (5,4))
plt.pie(market_share, labels = brands, autopct = '%1.1f%%', wedgeprops = dict(width = 0.6))
plt.text(0,0, 'Sales', ha = 'center')
plt.show()
```

# Multiple sub-charts 

```python
# method-1 : using plt.subplot() : create one plot at a time. good for less number of subplots
# method-2 : using plt.subplots() : create all plots at once. good for large number of subplots
```

# method-1 : using plt.subplot()

```python
Dream_breakfast_Thali = ["Idli", "Dosa", "Bonda", "Puri"]
share = [10, 40, 25, 25]

plt.figure(figsize = (8,7))

plt.subplot(2, 3, 1)
plt.pie(share, labels = Dream_breakfast_Thali)

plt.subplot(2, 3, 2)
plt.bar(Dream_breakfast_Thali, share)

plt.subplot(2, 3, 3)
plt.barh(Dream_breakfast_Thali, share)

plt.subplot(2, 3, 4)
plt.pie(share, labels = Dream_breakfast_Thali)

plt.subplot(2, 3, 5)
plt.barh(Dream_breakfast_Thali, share)

plt.subplot(2, 3, 6)
plt.pie(share, labels = Dream_breakfast_Thali)

plt.tight_layout()
plt.show()
```

# plt.subplots()

```python
Lunch_thali = ["Biryani", "Pulao", "Mutton", "fish", "coke"]
share = [50, 10, 20, 15, 5]

plt.figure(figsize = (20, 16))

# fig : whole image canvas
# axes : array of subplots ( 2 x 2)
fig, axes = plt.subplots(2,2)
titles = ["thali-1", "thali-2", "thali-3", "thali-4"]
for ax, t in zip(axes.ravel(), titles):  # ravel() convert to 1D
    if t == "thali-2":
        ax.bar(Lunch_thali,share)
        ax.set_title(t)
    else:
        ax.pie(share, labels = Lunch_thali, autopct = "%1.1f%%")
        ax.set_title(t)
        
plt.tight_layout()
```

# 3D charts

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data
# x,y,z are the corrdinates
z = np.linspace(0,10, 100)  # generating 0 to 10, 100 linear spaced elements
x = np.sin(z)  # this is generting data in sin shape (-1 to +1)
y = np.cos(z)  # this is generting data in cos shape (-1 to +1)

fig = plt.figure()
# 1 : row, 1 : col , 1 : 1st chart
ax = fig.add_subplot(111, projection = '3d')

ax.plot(x, y, z)  # plot() function is used for plotting line chart
```

```python
# 3D Scatter plot
# x,y,z are the corrdinates
# rand() function is generating 100 elements in 3D format
x, y, z = np.random.rand(3,100)

fig = plt.figure()
# 1 : row, 1 : col , 1 : 1st chart
ax = fig.add_subplot(111, projection = '3d')

ax.scatter(x, y, z)  # plot() function is used for plotting line chart
```

```python
# Data
# x,y,z are the corrdinates
x = np.linspace(-1,1, 30)  # generating 0 to 10, 100 linear spaced elements
y = np.linspace(-1,1, 30)
x_new, y_new = np.meshgrid(x,y)
z = np.sin(x_new + 10) * np.sin(y_new + 10)

fig = plt.figure()
# 1 : row, 1 : col , 1 : 1st chart
ax = fig.add_subplot(111, projection = '3d')

ax.plot_wireframe(x, y, z)  # plot() function is used for plotting line chart
```

```python
# ploting convex 3D chart
import plotly.graph_objects as go

x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)
x_new, y_new = np.meshgrid(x,y)

z = np.cos(x_new) + np.cos(y_new)

fig = go.Figure(data = [go.Surface(z = z, x = x_new, y = y_new)])
fig.show()
```
